```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# List your uploaded filenames
file_names = [
    "Peter Cat Recording Co.： “Memory Box” KCRW Live from HQ [K6TWekfnjDQ].wav"
]

for file in file_names:
    y, sr = librosa.load(file)

    # Compute MFCC
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)

    # Compute Mel spectrogram
    mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=256, fmax=8000)

    # Compute Chromagram
    chroma = librosa.feature.chroma_stft(y=y, sr=sr, n_chroma=12)

    # Compute Spectrogram
    S = librosa.amplitude_to_db(np.abs(librosa.stft(y))**2, ref=np.max)

    # Compute Zero-Crossing Rate
    zcr = librosa.feature.zero_crossing_rate(y)

    # 1. Plot and Analyze MFCC
    plt.figure(figsize=(18, 10), dpi=300)
    librosa.display.specshow(mfcc, x_axis='time', sr=sr)
    plt.colorbar()
    plt.title(f"MFCC: {file}", fontsize=16)
    plt.tight_layout(pad=5.0)
    plt.savefig(f"{file}_mfcc.png")
    plt.show()

    # Analysis: MFCCs represent the short-term power spectrum of sound and are commonly used in speech recognition and music classification.
    print(f"MFCC Analysis for {file}:")
    print(f"  - Number of MFCC coefficients: {mfcc.shape[0]}")
    print(f"  - Duration of the audio: {y.shape[0] / sr} seconds")
    print(f"  - Temporal evolution of the sound's spectral features is shown in the plot.\n")

    # 2. Plot and Analyze Mel Spectrogram
    plt.figure(figsize=(18, 10), dpi=300)
    librosa.display.specshow(librosa.power_to_db(mel_spec, ref=np.max), x_axis='time', y_axis='mel', sr=sr)
    plt.colorbar()
    plt.title(f"Mel Spectrogram: {file}", fontsize=16)
    plt.tight_layout(pad=5.0)
    plt.savefig(f"{file}_mel_spectrogram.png")
    plt.show()

    # Analysis: The Mel spectrogram shows the frequency content of the signal in the Mel scale, which is more aligned with human hearing.
    print(f"Mel Spectrogram Analysis for {file}:")
    print(f"  - Frequency range: {mel_spec.shape[0]} Mel bins")
    print(f"  - Temporal resolution: {mel_spec.shape[1]} frames")
    print(f"  - This representation is useful for analyzing speech and music, as the Mel scale mimics human auditory perception.\n")

    # 3. Plot and Analyze Chromagram
    plt.figure(figsize=(18, 10), dpi=300)
    librosa.display.specshow(chroma, x_axis='time', y_axis='chroma', sr=sr)
    plt.colorbar()
    plt.title(f"Chromagram: {file}", fontsize=16)
    plt.tight_layout(pad=5.0)
    plt.savefig(f"{file}_chromagram.png")
    plt.show()

    # Analysis: The chromagram is useful for music analysis as it represents the harmonic content of the audio, especially in terms of pitch classes (C, D, E, etc.).
    print(f"Chromagram Analysis for {file}:")
    print(f"  - Number of pitch classes: {chroma.shape[0]} (usually 12 for Western music)")
    print(f"  - This feature is used to detect chord progressions and tonal centers in music.\n")

    # 4. Plot and Analyze Spectrogram
    plt.figure(figsize=(18, 10), dpi=300)
    librosa.display.specshow(S, x_axis='time', y_axis='log', sr=sr)
    plt.colorbar()
    plt.title(f"Spectrogram: {file}", fontsize=16)
    plt.tight_layout(pad=5.0)
    plt.savefig(f"{file}_spectrogram.png")
    plt.show()

    # Analysis: The spectrogram shows the full spectrum of frequencies over time. It's useful for analyzing both transient and sustained elements of sound.
    print(f"Spectrogram Analysis for {file}:")
    print(f"  - This representation captures both high and low frequencies over time.")
    print(f"  - The log-scale y-axis highlights the frequencies with more perceptual detail at lower frequencies.\n")

    # 5. Plot and Analyze Zero-Crossing Rate
    plt.figure(figsize=(18, 10), dpi=300)
    plt.plot(np.linspace(0, len(y)/sr, len(zcr[0])), zcr[0], label="Zero-Crossing Rate")
    plt.title(f"Zero Crossing Rate: {file}", fontsize=16)
    plt.xlabel('Time (s)')
    plt.ylabel('Rate')
    plt.tight_layout(pad=5.0)
    plt.savefig(f"{file}_zero_crossing_rate.png")
    plt.show()

    # Analysis: The zero-crossing rate represents the number of times the signal crosses the zero amplitude axis, which is useful for distinguishing between percussive and tonal sounds.
    print(f"Zero-Crossing Rate Analysis for {file}:")
    print(f"  - This feature is often used for audio segmentation, especially in speech or music where percussive events can be isolated.\n")
```

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

file = "Kacey Musgraves - Follow Your Arrow (Official Music Video) [kQ8xqyoZXCc].wav"

y, sr = librosa.load(file)

# 1. MFCC
mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
plt.figure(figsize=(14, 6), dpi=150)
librosa.display.specshow(mfcc, x_axis='time', sr=sr)
plt.colorbar()
plt.title("MFCC - Mel Frequency Cepstral Coefficients", fontsize=16)
plt.tight_layout()
plt.show()
print("MFCC Analysis:\nMFCCs capture the timbral texture of the song. You can observe smooth transitions, indicating a melodic and consistent vocal delivery common in country-pop music.\n")

# 2. Mel Spectrogram
mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128, fmax=8000)
plt.figure(figsize=(14, 6), dpi=150)
librosa.display.specshow(librosa.power_to_db(mel_spec, ref=np.max), x_axis='time', y_axis='mel', sr=sr)
plt.colorbar(format='%+2.0f dB')
plt.title("Mel Spectrogram", fontsize=16)
plt.tight_layout()
plt.show()
print("Mel Spectrogram Analysis:\nShows energy distribution across frequency bands. Strong energy below 4000Hz suggests emphasis on vocals and acoustic instrumentation.\n")

# 3. Chromagram
chroma = librosa.feature.chroma_stft(y=y, sr=sr)
plt.figure(figsize=(14, 6), dpi=150)
librosa.display.specshow(chroma, x_axis='time', y_axis='chroma', sr=sr)
plt.colorbar()
plt.title("Chromagram", fontsize=16)
plt.tight_layout()
plt.show()
print("Chromagram Analysis:\nHighlights harmonic content. Distinct vertical lines suggest chord changes and progression, characteristic of structured songwriting.\n")

# 4. Spectrogram (Log Frequency)
S = librosa.amplitude_to_db(np.abs(librosa.stft(y))**2, ref=np.max)
plt.figure(figsize=(14, 6), dpi=150)
librosa.display.specshow(S, x_axis='time', y_axis='log', sr=sr)
plt.colorbar(format='%+2.0f dB')
plt.title("Spectrogram (Log Frequency)", fontsize=16)
plt.tight_layout()
plt.show()
print("Spectrogram Analysis:\nShows detailed time-frequency information. Moderate high-frequency content appears due to cymbals or guitar strumming.\n")

# 5. Zero Crossing Rate
zcr = librosa.feature.zero_crossing_rate(y)
plt.figure(figsize=(14, 4), dpi=150)
plt.plot(np.linspace(0, len(y)/sr, len(zcr[0])), zcr[0], label="ZCR")
plt.title("Zero-Crossing Rate", fontsize=16)
plt.xlabel("Time (s)")
plt.ylabel("Rate")
plt.tight_layout()
plt.show()
print("ZCR Analysis:\nLow ZCR throughout most of the song suggests smooth, non-percussive sounds, fitting for vocals and acoustic instrumentation.\n")
```

```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

# Load the audio file
file = "Taylor Swift - State Of Grace (Taylor's Version) (Lyric Video) [-mrC5tRkxrY].wav"
y, sr = librosa.load(file)

# Compute features
mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=256, fmax=8000)
chroma = librosa.feature.chroma_stft(y=y, sr=sr, n_chroma=12)
S = librosa.amplitude_to_db(np.abs(librosa.stft(y))**2, ref=np.max)
zcr = librosa.feature.zero_crossing_rate(y)

# --- MFCC ---
plt.figure(figsize=(12, 4), dpi=150)
librosa.display.specshow(mfcc, x_axis='time', sr=sr)
plt.colorbar()
plt.title("MFCC - Captures the timbral texture of the song")
plt.tight_layout()
plt.show()

# Analysis
print("MFCC Analysis: Reveals how timbre evolves. The smooth flow indicates consistent vocal/instrumental tones with variation during chorus or bridge.")

# --- Mel Spectrogram ---
plt.figure(figsize=(12, 4), dpi=150)
librosa.display.specshow(librosa.power_to_db(mel_spec, ref=np.max), x_axis='time', y_axis='mel', sr=sr)
plt.colorbar()
plt.title("Mel Spectrogram - Energy distribution across frequencies")
plt.tight_layout()
plt.show()

print("Mel Spectrogram Analysis: Highlights frequency energy. Bright zones suggest intense vocals or instruments. Darker areas denote softer segments.")

# --- Chromagram ---
plt.figure(figsize=(12, 4), dpi=150)
librosa.display.specshow(chroma, x_axis='time', y_axis='chroma', sr=sr)
plt.colorbar()
plt.title("Chromagram - Harmonic and pitch class representation")
plt.tight_layout()
plt.show()

print("Chromagram Analysis: Shows tonal structure and key transitions. Stable bands suggest steady harmonic content.")

# --- Spectrogram ---
plt.figure(figsize=(12, 4), dpi=150)
librosa.display.specshow(S, x_axis='time', y_axis='log', sr=sr)
plt.colorbar()
plt.title("Spectrogram - Log-scaled frequency energy over time")
plt.tight_layout()
plt.show()

print("Spectrogram Analysis: Captures full spectrum detail. Useful for identifying percussive hits or sustained instruments.")

# --- Zero-Crossing Rate ---
plt.figure(figsize=(12, 3), dpi=150)
plt.plot(np.linspace(0, len(y)/sr, len(zcr[0])), zcr[0], label="Zero-Crossing Rate")
plt.title("Zero Crossing Rate - Measures noisiness or percussiveness")
plt.xlabel("Time (s)")
plt.ylabel("Rate")
plt.tight_layout()
plt.show()

print("Zero Crossing Rate Analysis: Peaks indicate transient or noisy moments. Useful for rhythm or percussive detection.")
```
